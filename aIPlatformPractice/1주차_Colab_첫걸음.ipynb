### 셀은 '순서대로', 변수는 '기억됩니다'
Colab은 셀을 실행할 때마다 그 결과를 **기억**해 둡니다.
그래서 앞 셀에서 만든 값을 뒤 셀에서 그대로 쓸 수 있습니다.
반대로 **앞 셀을 실행하지 않고 뒤 셀만 실행하면 오류**가 납니다. 직접 확인해 봅시다.


````python
# ============================================================
# [실습 1-2] 값을 상자에 담아두기 (변수)
# ------------------------------------------------------------
# '변수'는 값을 담아두는 이름표 붙은 상자라고 생각하면 쉽습니다.
# = 기호는 "같다"가 아니라 "오른쪽 값을 왼쪽 상자에 넣어라"라는 뜻입니다.
# ============================================================

my_name = "홍길동"   # 문자열(글자)을 my_name 이라는 상자에 담습니다
my_grade = 1         # 정수(숫자)를 my_grade 상자에 담습니다

print("변수에 값을 저장했습니다.")
````

````python
# ============================================================
# [실습 1-3] 앞 셀에서 만든 변수를 여기서 사용하기
# ------------------------------------------------------------
# 바로 위 셀([실습 1-2])을 실행했다면 아래 코드가 잘 동작합니다.
# 만약 위 셀을 실행하지 않았다면
#   NameError: name 'my_name' is not defined
# 라는 오류가 납니다. -> "그런 이름의 상자가 없다"는 뜻입니다.
# ============================================================

# f"..." 형식은 문자열 안에 { } 로 변수 값을 끼워 넣는 방법입니다.
print(f"제 이름은 {my_name}이고, {my_grade}학년입니다.")
````

> 💡 **여기서 꼭 기억할 것**

# STEP 2. 파이썬 기초 되짚기  (약 15분)

이번 학기 내내 쓸 최소한의 문법만 빠르게 확인합니다.
이미 아는 내용이면 실행만 하고 넘어가도 좋습니다.


````python
# ============================================================
# [실습 2-1] 자료형 — 값의 '종류'
# ------------------------------------------------------------
# type( ) 은 그 값이 어떤 종류인지 알려주는 명령입니다.
#   str   = 문자열(글자)   int = 정수      float = 소수      bool = 참/거짓
# ============================================================

course = "인공지능 플랫폼실습"   # str  : 따옴표로 감싼 글자
week = 1                        # int  : 소수점 없는 숫자
score = 4.5                     # float: 소수점 있는 숫자
is_fun = True                   # bool : True(참) 또는 False(거짓)

print(course, "->", type(course))
print(week,   "->", type(week))
print(score,  "->", type(score))
print(is_fun, "->", type(is_fun))
````

````python
# ============================================================
# [실습 2-2] 리스트와 반복문
# ------------------------------------------------------------
# 리스트는 여러 값을 순서대로 담는 상자입니다. 대괄호 [ ] 를 씁니다.
# for 문은 리스트 안의 값을 하나씩 꺼내서 반복 실행합니다.
# ============================================================

# 이번 학기에 배울 플랫폼들을 리스트에 담아봅니다
platforms = ["Colab", "Kaggle", "Hugging Face", "Gradio", "Streamlit"]

print("총 개수:", len(platforms))   # len( ) 은 개수를 세는 명령
print("첫 번째:", platforms[0])     # 파이썬은 0번부터 셉니다 (0, 1, 2 ...)
print("마지막:", platforms[-1])     # -1 은 뒤에서 첫 번째

print("\n--- 하나씩 꺼내보기 ---")
# enumerate( ) 를 쓰면 '순번'과 '값'을 함께 꺼낼 수 있습니다.
# start=1 은 순번을 1부터 시작하라는 뜻입니다.
for number, name in enumerate(platforms, start=1):
    print(f"{number}번째 플랫폼: {name}")
````

````python
# ============================================================
# [실습 2-3] 조건문 — 상황에 따라 다르게 실행하기
# ------------------------------------------------------------
# if / elif / else 로 조건을 나눕니다.
# ★ 중요: if 줄 끝에는 콜론( : )을 쓰고, 다음 줄은 반드시 '들여쓰기(띄어쓰기 4칸)'합니다.
#         들여쓰기를 빠뜨리면 IndentationError 오류가 납니다.
# ============================================================

my_score = 85

if my_score >= 90:
    grade = "A"
elif my_score >= 80:      # 위 조건이 아닐 때 검사
    grade = "B"
else:                     # 위 조건이 모두 아닐 때
    grade = "C"

print(f"점수 {my_score}점 -> 등급 {grade}")
````

````python
# ============================================================
# [실습 2-4] 함수 — 자주 쓰는 코드에 이름 붙이기
# ------------------------------------------------------------
# def 함수이름(재료):  형태로 만들고, return 으로 결과를 돌려줍니다.
# 앞으로 AI 모델을 부를 때도 결국 '함수를 호출'하는 형태입니다.
# ============================================================

def introduce(name, grade):
    """이름과 학년을 받아서 소개 문장을 만들어 돌려주는 함수"""
    message = f"안녕하세요, {grade}학년 {name}입니다."
    return message          # 만든 문장을 돌려줍니다

# 함수는 '정의'만 하면 아무 일도 일어나지 않습니다. 반드시 '호출'해야 실행됩니다.
result = introduce("홍길동", 1)
print(result)

# 재료를 바꿔서 여러 번 호출할 수 있는 것이 함수의 장점입니다.
print(introduce("김철수", 1))
````

# STEP 4. 그래프 그려보기 + 한글 깨짐 해결  (약 10분)

앞으로 실습 결과를 그래프로 자주 확인합니다.
그런데 Colab은 기본 상태에서 **한글이 네모(□□□)로 깨져서** 나옵니다.
이번에 해결 방법까지 같이 익혀 둡시다.


````python
# ============================================================
# [실습 4-1] 기본 그래프 그리기 (영문)
# ------------------------------------------------------------
# matplotlib 는 그래프를 그리는 대표 도구 상자입니다.
# 관례적으로 plt 라는 짧은 별명을 붙여서 사용합니다.
# ============================================================

import matplotlib.pyplot as plt

weeks  = [1, 2, 3, 4, 5]                 # x축 값
scores = [60, 68, 75, 84, 92]            # y축 값

plt.figure(figsize=(5, 3))               # 그래프 크기 (가로, 세로)
plt.plot(weeks, scores, marker="o")      # 선 그래프, marker="o" 는 점 표시
plt.title("My AI Skill")                 # 제목
plt.xlabel("Week")                       # x축 이름
plt.ylabel("Score")                      # y축 이름
plt.grid(True)                           # 격자 표시
plt.savefig("step4_english.png", dpi=100, bbox_inches="tight")
plt.show()                               # 화면에 그리기

print("그래프가 보이면 성공입니다.")
````

````python
# ============================================================
# [실습 4-2] 한글 폰트 설치 + 적용  ※ 실행에 1~2분 걸립니다
# ------------------------------------------------------------
# ★ 설치와 적용을 '한 셀'에 함께 넣었습니다.
#   따로 나누면 설치 셀을 건너뛰고 실행해서 아래 오류가 납니다.
#     FileNotFoundError: [Errno 2] No such file or directory:
#       '/usr/share/fonts/truetype/nanum/NanumGothic.ttf'
#   -> "그런 파일이 없다" = 폰트가 아직 설치되지 않았다는 뜻입니다.
#
# 이 셀이 하는 일
#   1) 나눔 폰트를 설치한다
#   2) 설치된 폰트 파일의 위치를 '직접 찾아낸다'  <- 경로가 달라져도 안전
#   3) matplotlib에 등록하고 기본 글꼴로 지정한다
# ============================================================

# 1) 폰트 설치   ( -y : 질문에 자동으로 '예' / -qq : 메시지 최소화 )
!apt-get install -y -qq fonts-nanum

# 2) 설치된 나눔 폰트 파일을 직접 찾기
import glob                              # 파일 이름 패턴으로 파일을 찾는 도구
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

# /usr/share/fonts 아래를 전부 뒤져서 이름이 Nanum 으로 시작하는 ttf 파일을 찾습니다.
#   ** 와 recursive=True 는 "하위 폴더까지 전부" 라는 뜻입니다.
font_files = glob.glob("/usr/share/fonts/**/Nanum*.ttf", recursive=True)
print("찾은 나눔 폰트 개수 :", len(font_files))

if len(font_files) == 0:
    # 설치가 덜 끝났거나 실패한 경우
    print("❌ 폰트를 찾지 못했습니다. 이 셀을 한 번 더 실행해 보세요.")
    print("   그래도 안 되면 [런타임] -> [모두 실행] 후 다시 시도하세요.")
else:
    # NanumGothic.ttf 가 있으면 그것을, 없으면 찾은 것 중 첫 번째를 사용합니다.
    FONT = ([p for p in font_files if p.endswith("NanumGothic.ttf")]
            or sorted(font_files))[0]

    fm.fontManager.addfont(FONT)                   # matplotlib에 폰트 등록
    FONT_NAME = fm.FontProperties(fname=FONT).get_name()
    plt.rcParams["font.family"] = FONT_NAME        # 기본 글꼴로 지정
    plt.rcParams["axes.unicode_minus"] = False     # 마이너스(-) 기호 깨짐 방지

    print("✅ 준비 완료")
    print("사용 폰트 :", FONT_NAME)
    print("폰트 경로 :", FONT)
````

````python
# ============================================================
# [실습 4-3] 한글 제목으로 그래프 다시 그리기
# ------------------------------------------------------------
# ★ 위 [실습 4-2] 셀을 먼저 실행해야 합니다.
#   많은 자료가 "런타임을 재시작하라"고 안내하지만,
#   4-2처럼 폰트를 직접 등록하면 재시작 없이 바로 한글이 나옵니다.
# ============================================================

import matplotlib.pyplot as plt

weeks  = [1, 2, 3, 4, 5]
scores = [60, 68, 75, 84, 92]

plt.figure(figsize=(5, 3))
plt.plot(weeks, scores, marker="o")
plt.title("나의 AI 실력 변화")     # 한글 제목
plt.xlabel("주차")
plt.ylabel("점수")
plt.grid(True)
plt.savefig("step4_korean.png", dpi=100, bbox_inches="tight")
plt.show()

print("제목이 □□□ 가 아니라 한글로 보이면 성공입니다.")
````

# 오늘 자주 나온 오류와 해결법

| 증상(오류 메시지) | 원인 | 해결 방법 |
|---|---|---|
| `NameError: name '...' is not defined` | 위쪽 셀을 실행하지 않음 | 런타임 → 모두 실행(Run all) |
| `IndentationError` | 들여쓰기가 안 맞음 | `if`, `for`, `def` 다음 줄은 띄어쓰기 4칸 |
| `SyntaxError` | 따옴표·괄호를 빼먹음 | 빨간 밑줄 위치 앞뒤를 확인 |
| 그래프 한글이 □□□ | 한글 폰트 미설치 | STEP 4-2 를 실행 |
| `FileNotFoundError: ... NanumGothic.ttf` | 폰트 설치 전에 폰트를 등록하려 함 | [실습 4-2] 셀 **전체**를 처음부터 실행 |
| `GPU 사용 가능 : False` | GPU 런타임 미설정 | 런타임 → 런타임 유형 변경 → T4 GPU |
| GPU를 켜도 배정 안 됨 | 무료 등급은 GPU 보장 없음 | 시간을 두고 재시도 / CPU로 진행 |
| 실행이 갑자기 멈추고 변수가 사라짐 | 세션 종료(장시간 방치) | 런타임 → 모두 실행으로 다시 실행 |
| 드라이브 연결이 엉뚱한 계정으로 됨 | 구글 계정 여러 개 로그인 | 브라우저 시크릿 창에서 다시 시도 |

> 🙋 **막혔을 때 행동 요령**
> ① 오류 메시지의 **맨 아랫줄**을 먼저 읽기 → ② 위 표에서 찾아보기
> → ③ 옆 사람 화면과 비교하기 → ④ 그래도 안 되면 손 들어 강사 호출
