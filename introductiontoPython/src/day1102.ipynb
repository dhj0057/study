```python
import matplotlib.pyplot as plt

year = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022]
gdp = [3070.9, 3250.1, 3398.8, 3574, 3678.2, 3721.8, 3744, 4033.6, 4165.5]
jeju = [15.6, 16.5, 15.7, 15.6, 16.2, 16.8, 16.7, 17.5, 17, 15.6]

plt.figure(figsize = (5, 3))
plt.plot(year, gdp, 'c.-')
plt.title('GDP  per capita')
plt.xlabel('years')
plt.ylabel("gdp")
plt.ylim(2500, 4500)
plt.show()
```

```python
import matplotlib.pyplot as plt
year = [2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022]
seoul = [12.1,12,12.2,12.5,13.4,13.6,13.6,13,12.9,13.5,13.2,13.7,13.2]
jeju = [15.6,15.6,15.7,16.5,16.2,16.7,17,16.8,16.6,16.8,16.7,17.5,17]

plt.figure(figsize = (5, 3))
plt.plot(year, jeju, label = '제주')
plt.plot(year, seoul, label = '서울')
plt.ylim(10, 22)
plt.title('온도')
plt.xlabel('연도')
plt.ylabel("C", rotation = 0, loc = 'top')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

name = ['홍길동', '고길동', '신길동']
x = [75, 92, 83]
y = [79, 89, 90]

plt.figure(figsize=(5,3))
plt.scatter(x, y)
for i in range(len(name)):
  plt.annotate(name[i],xy=(x[i],y[i]),xytext=(5,-10),textcoords='offset points')
plt.xlim(73, 95)
plt.ylim(75, 93)
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```
